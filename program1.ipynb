```python
import numpy as np
import pandas as pd
```

```python
a = np.array([10, 20, 30, 40, 50])
```

```python
print("Array:", a)
print("Mean:", a.mean())
print("Reshaped 2D:\n", a.reshape(5, 1))
print("Sliced:", a[1:4])
```

```python
data = {
    "Name": ["Asha", "Rani", "Meera", "Karan"],
    "Marks": [88, 72, 91, 65],
    "Branch": ["CSE", "ECE", "CSE", "ISE"]
}
```

```python
df = pd.DataFrame(data)
```

```python
print(df)

print("\nFirst 2 rows:\n", df.iloc[0:2])

print("\nMarks column:\n", df["Marks"])

print("\nRows where marks > 70:\n", df[df["Marks"] > 70])
```
